Nested if

```python
#checking eligibility for voting and citizenship.
age = 20
citizen = "no"
if age >= 18:
    if citizen == "yes":
        print("you are eligible to vote")
    else: 
        print("you must be a citizen to vote")
else:
        print("you must be 18 or older to vote")
```

```python
#Check if a number is positive and even 
num = int(input("Enter a number:"))
if num > 0:
    print("It's positive")
    if num % 2 == 0:
        print("It's even")
    else:
        print("It's odd")
else:
    print("It's not positive")
```

```python
a = 20 # age
b = True # Citizen status
if a >= 18 and b==True:
    print("Eligible")
else:
    print("Ineligible")
```

```python
#Check if a person can get a driving license 
age = int(input("Enter your age: "))
has_license = input("Do you have a learner's license? (yes/no):")
if age >= 18 and has_license =="yes":
    print("you can drive")
else:
    print("you are not eligible yet")
```
